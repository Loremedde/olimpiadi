# Analisi dei dati - Olimpiadi Milano Cortina

```python
import pandas as pd

dati = pd.read_csv("interventi.csv")
dati.columns = [c.strip() for c in dati.columns]

tot_interventi = len(dati)
fasi = dati["Fase dell'intervento"].value_counts().to_dict()
totale_economico = dati["Totale quadro economico"].sum()
media_economica = dati["Totale quadro economico"].mean()
per_area = dati["Territorio"].value_counts().to_dict()
assegnatario_vuoto = (dati["Aggiudicatario"] == "---").mean() * 100
prima_olimpiadi = (dati["fine ante olimpiadi"] == "sì").mean() * 100 if "fine ante olimpiadi" in dati.columns else None

ordine_fasi = ["in progettazione", "in gara", "in esecuzione", "concluso"]
dati["fase_index"] = dati["Fase dell'intervento"].apply(lambda f: ordine_fasi.index(f) if f in ordine_fasi else None)
media_fase_index = dati["fase_index"].mean()

report = {
    "Totale interventi": tot_interventi,
    "Totale quadro economico": round(totale_economico, 2),
    "Media economica": round(media_economica, 2),
    "Distribuzione per fase": fasi,
    "Distribuzione per territorio": per_area,
    "Percentuale senza assegnatario": round(assegnatario_vuoto, 2),
    "Percentuale con fine ante Olimpiadi": round(prima_olimpiadi, 2) if prima_olimpiadi is not None else "N/D",
    "Indice medio avanzamento (0=inizio,3=fine)": round(media_fase_index, 2)
}

for k, v in report.items():
    print(f"{k}: {v}")
```
